```python
from sklearn.datasets import load_iris
from sklearn.metrics import classification_report
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

iris = load_iris()
x, test, y, labels = train_test_split(
    iris.data, iris.target, test_size=0.3, stratify=iris.target, random_state=7
)
model = make_pipeline(
    StandardScaler(), MLPClassifier(hidden_layer_sizes=(16, 8), solver="lbfgs", random_state=7)
)
model.fit(x, y)
print(classification_report(labels, model.predict(test), target_names=iris.target_names))
```
